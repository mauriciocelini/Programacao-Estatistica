## Exercício 1
Crie um vetor com 20 números inteiros aleatórios entre 0 e 100.  
- Mostre o vetor.  
- Calcule média, desvio-padrão e valor máximo.  

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
vec = np.random.uniform(low=0, high=100, size=20)
media = np.mean(vec)
dp = np.std(vec)
maximo = np.max(vec)

print(f'O vetor tem os seguintes valores: \n {vec}')
print(f'\nEsse vetor de valores tem como média: {media}, desvio padrão: {dp} e valor máximo {maximo}')
```

## Exercício 2
Gere uma matriz $5 \times 5$ com valores da distribuição normal padrão.  
- Calcule a soma de cada linha e de cada coluna.  
- Extraia a diagonal principal.  

```python
matriz = np.random.normal(loc=0, scale=1, size=(5, 5))
print(matriz)
print()

soma_colunas = matriz.sum(axis=0)
print(f'A soma das colunas é {soma_colunas}')
print()

soma_linhas = matriz.sum(axis=1)
print(f'A soma das linhas é {soma_linhas}')
print()

diagonal = np.diag(matriz)
print(f'Os valores da diagonal da matriz são {diagonal}')
```

## Exercício 3
Implemente uma função que recebe dois vetores NumPy e retorna:  
- Produto escalar (`np.dot`).  
- Distância euclidiana entre eles (`np.linalg.norm`).  

```python
def calculos_vetores(vetor1,vetor2):
    prod_escalar = np.dot(vetor1, vetor2)
    distancia = np.linalg.norm(vetor1 - vetor2)
    return prod_escalar, distancia

v1 = np.array([1,2,3])
v2 = np.array([4,5,6])
produto_escalar, distancia_euclidiana = calculos_vetores(v1,v2)
print(produto_escalar)
```

## Exercício 4
Crie uma matriz quadrada $A \in \mathbb{R}^{4 \times 4}$ com valores inteiros aleatórios entre $-5$ e $5$.  
- Verifique se $A$ é invertível calculando $\det(A)$ (`np.linalg.det`).  
- Caso seja, calcule $A^{-1}$ (`np.linalg.inv`) e verifique se $A \cdot A^{-1} = I$.  

```python
matriz = np.random.uniform(-5, 5, size=(4, 4))
print(matriz)
print()

determinante = np.linalg.det(matriz)
print(determinante)
print()

matriz_inversa = np.linalg.inv(matriz)
print(matriz_inversa)
print()

multi = matriz @ matriz_inversa
print(multi)
print()
```

## Exercício 5
Gere uma matriz simétrica $B \in \mathbb{R}^{5 \times 5}$ a partir de uma matriz aleatória $M$ definida por $B = \tfrac{1}{2}(M + M^T)$.  
- Calcule os autovalores e autovetores (`np.linalg.eig`).  
- Verifique numericamente a relação $Bv = \lambda v$.  

```python
matriz = np.random.uniform(size=(5,5))
B = 0.5*(matriz + matriz.T)
print(B)
print()

autovalores, autovetores = np.linalg.eig(B)
autovalores = autovalores.real
autovetores = autovetores.real
print(autovalores)
print()
print(autovetores)
print()

# 1. Pega o primeiro autovalor e o seu autovetor (1ª coluna)
lam = autovalores[0]
v = autovetores[:, 0]

# 2. Calcula os dois lados da relação Bv = lambda * v
lado_esquerdo = B @ v
lado_direito = lam * v

print("--- Lado Esquerdo (B @ v) ---:\n", lado_esquerdo, "\n")
print("--- Lado Direito (lambda * v) ---:\n", lado_direito, "\n")

# 3. Compara se os dois resultados são numericamente equivalentes
sao_iguais = np.allclose(lado_esquerdo, lado_direito)
print("--- A relação é verdadeira? ---:\n", sao_iguais)
```
