# Exercício 1

Utilizando o método ingênuo gere uma amostra de tamanho $M=1000$ de uma variável geométrica com parâmetro $p$. Compara o gráfico da sua distribuição com a função do numpy. Além disso, compare o que acontece quando você varia o valor $p$

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
```

```python
M = 1000
amostra = []
p = 0.01
for i in range(M):
    bern = np.random.uniform() < p 
    x = 1
    while bern != 1:
        bern = np.random.uniform() < p
        x += 1
    amostra.append(x)
```

Aqui eu utilizei as bernoullis para gerar uma geometrica, e gerei uma amostra de várias geométricas

```python
plt.hist(amostra)
```

* Gerando uma amostra de geométricas

```python
N = 10000
p = 0.1
amostra = []
for i in range(1,N+1):
    u = np.random.uniform()
    x = np.floor(np.log(u)/np.log(1-p)) + 1
    amostra.append(x)
```
