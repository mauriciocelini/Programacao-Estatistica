### Exercício 1 — Estimação de integrais

Considere a integral

$$
I=\int_0^1 e^{-x^2}\,dx.
$$

Utilizando $X\sim U(0,1)$, estime $I$ por Monte Carlo, calcule a variância do estimador e construa um intervalo de confiança de 95%.

```python
unif = np.random.uniform(size=n)
esp = np.exp(-unif**2)
integral = np.mean(esp)
print(integral)
```

```python
var_estimado = np.var(esp, ddof=1)
print(var_estimado)
```

```python
margem = 1.96*np.sqrt(var_estimado/n)
IC_95 = (integral - margem, integral + margem)
```

### Exercício 3 — Estimação de $\pi$

Considere $(X,Y)$ uniformemente distribuído em $[0,1]^2$ e o evento

$$
A=\{X^2+Y^2\leq 1\}.
$$

Utilize Monte Carlo para estimar $\pi$, calcule a variância do estimador e construa um intervalo de confiança de 95%.

```python
unif_x = np.random.uniform(size=n)
unif_y = np.random.uniform(size=n)
amostra_pi = [(unif_x**2 + unif_y**2 < 1)*4]
print(np.mean(amostra_pi))
var_estimada = np.var(amostra_pi, ddof=1)
print(var_estimada)
```
