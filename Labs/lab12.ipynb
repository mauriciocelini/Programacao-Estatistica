### Normal

```python
def densidade_normal(x, mi, sigma2):
    return (2/np.sqrt(2*np.pi*sigma2))*np.exp(-1/(2*sigma2)*(x-mi)**2)
```

```python
mis = [-2,-1,0,1,2]
x = np.linspace(-6,6, num=200)

for mi in mis:
    plt.plot(x, densidade_normal(x,mi,1), label=rf'$\mu$ = {mi}')
plt.legend()
plt.show()
```

### Exponencial

```python
def densidade_exp(x, lamb):
    return lamb*np.exp(-lamb*x)
```

```python
lambs = [2, 1.5, 1, 0.5, 0.25] # a inclinada é lamb - t 
y = np. linspace(0, 7)

for lamb in lambs:
    plt.plot(y,densidade_exp(y,lamb), label=rf'$\lambda$ = {lamb}' )
plt.legend()
plt.show()
```
