# Exercício 1

Utilizando o método ingênuo gere uma amostra de tamanho $M=1000$ de uma variável geométrica com parâmetro $p$. Compara o gráfico da sua distribuição com a função do numpy. Além disso, compare o que acontece quando você varia o valor $p$

```python
M = 1000
amostra = []
p = 0.01
for i in range(M):
    bern = np.random.uniform() < p 
    x = 1
    while bern != 1:
        bern = np.random.uniform() < p
        x += 1
    amostra.append(x)
```

Aqui eu utilizei as bernoullis para gerar uma geometrica, e gerei uma amostra de várias geométricas

```python
plt.hist(amostra)
```

```python
geometrica = np.random.geometric(p=0.01, size=1000)
```

Aqui eu apenas usei o np.geometric para gerar uma amostra de tamanho 100 de uma geométrica

```python
plt.hist(geometrica)
```

### Exercício 3

Gere uma amostra de tamanho 
 de uma Poisson com parâmetro 
 usando o método recursivo

```python
def pois(lamb):
    u = np.random.uniform()
    i = 0
    p = np.exp(-lamb)
    F = p
    while u > F:
        i += 1
        p *= lamb/i 
        F += p
    return i
```
