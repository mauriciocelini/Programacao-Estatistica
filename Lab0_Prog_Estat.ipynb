```python
distribuicoes = {
    "Normal": {"params": {"loc": 0, "scale": 1}, "dados": None, "estatisticas": None},
    "Exponencial": {"params": {"scale": 2}, "dados": None, "estatisticas": None},
    "Uniforme": {"params": {"low": -1, "high": 1}, "dados": None, "estatisticas": None}
}
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Aqui eu gerei os dados das minhas distribuições (de tamanho 1000 e parametros dependendo da distribuição) e adicionei nos respectivos dicionários

```python
distribuicoes["Normal"]["dados"] = np.random.normal(loc=0, scale=1, size=1000)
```

```python
distribuicoes["Exponencial"]["dados"] = np.random.exponential(scale=2, size=1000)
```

```python
distribuicoes["Uniforme"]["dados"] = np.random.uniform(low=-1, high=1, size=1000)
```

```python
#np.mean()
#np.median()
#np.std()
#np.var()
```

Aqui eu calculei algumas estatísticas descritivas (média, mediana, desvio padrão e variância), criei um dicionário com elas e adicionei como valor na chave "estatisticas" dentro de cada respectivo dicionário com sua respectiva  distribuição

```python
#valores para a normal

media_normal = np.mean(distribuicoes["Normal"]["dados"])
mediana_normal = np.median(distribuicoes["Normal"]["dados"])
desvio_padrao_normal = np.std(distribuicoes["Normal"]["dados"])
variancia_normal = np.var(distribuicoes["Normal"]["dados"])
distribuicoes["Normal"]["estatisticas"] = {"média": media_normal, "mediana": mediana_normal, "desvio_padrão":desvio_padrao_normal, "variância": variancia_normal}

```

```python
media_exp = np.mean(distribuicoes["Exponencial"]["dados"])
mediana_exp = np.median(distribuicoes["Exponencial"]["dados"])
desvio_padrao_exp = np.std(distribuicoes["Exponencial"]["dados"])
variancia_exp = np.var(distribuicoes["Exponencial"]["dados"])
distribuicoes["Exponencial"]["estatisticas"] = {"média": media_exp, "mediana": mediana_exp, "desvio_padrão":desvio_padrao_exp, "variância": variancia_exp}
```

```python
media_uni = np.mean(distribuicoes["Uniforme"]["dados"])
mediana_uni = np.median(distribuicoes["Uniforme"]["dados"])
desvio_padrao_uni = np.std(distribuicoes["Uniforme"]["dados"])
variancia_uni = np.var(distribuicoes["Uniforme"]["dados"])
distribuicoes["Uniforme"]["estatisticas"] = {"média": media_uni, "mediana": mediana_uni, "desvio_padrão":desvio_padrao_uni, "variância": variancia_uni}
```

Aqui eu plotei histogramas com os dados de cada distribuição, que no caso da normal por exemplo estão em distribuicoes["Normal"]["dados"]

```python
plt.hist(distribuicoes["Normal"]["dados"])
```

```python
plt.hist(distribuicoes["Exponencial"]["dados"])
```

```python
plt.hist(distribuicoes["Uniforme"]["dados"])
```

Aqui eu criei um DataFrame (uma tabela) que exibe as estatísticas descritivas que eu calcule para cada distribuição, para assim poder comparar os valores amostrais com os valores globais (populacionais)

```python
pd.DataFrame({"Normal": distribuicoes["Normal"]["estatisticas"], "Exponencial" : distribuicoes["Exponencial"]["estatisticas"], "Uniforme": distribuicoes["Uniforme"]["estatisticas"]})
```

Para a normal, a média amostral foi -0.0312, que é muito proximo da media populacional, como esperamos, ocorre o mesmo para a mediana
No desvio padrão acontece o mesmo, pois os valores amostrais sao muito proximos de 1, como a variancia populacional

Para a exponencial, o parametro verdadeiro (2), esta muito proximo da media amostral (1.97)

para a uniforme, a media esta dando muito proximo de 0, que é o ponto medio entre o valor maximo e minimo da uniforme

## Agora vamos adicionar ao nosso dicionário informacoes sobre outras distribuições (Poisson e Binomial)

Aqui eu apenas criei as chaves para as novas distribuicoes dentro do nosso dicionario "distribuicoes" e o valor (que sao outros dicionarios)

```python
distribuicoes["Poisson"] = {"params": {"lam": 5}, "dados" : None, "estatisticas" : None }
distribuicoes["Binomial"] = {"params": {"n": 1, "p" : 0.5}, "dados" : None, "estatisticas" : None}
```

Aqui eu gerei os dados e adicionei nos "dados" do meu dicionario

```python
distribuicoes["Poisson"]["dados"] = np.random.poisson(lam=5,size=1000)
```

```python
distribuicoes["Binomial"]["dados"] = np.random.binomial(n=1,p=0.5,size=1000)
```

Aqui eu calculei as novas descritivas e adicionei tambem no dicionario das distribuicoes e tambem atualizei o data frame

```python
media_pois = np.mean(distribuicoes["Poisson"]["dados"])
mediana_pois = np.median(distribuicoes["Poisson"]["dados"])
desvio_padrao_pois = np.std(distribuicoes["Poisson"]["dados"])
variancia_pois = np.var(distribuicoes["Poisson"]["dados"])
distribuicoes["Poisson"]["estatisticas"] = {"média": media_pois, "mediana": mediana_pois, "desvio_padrão":desvio_padrao_pois, "variância": variancia_pois}
```

```python
media_bin = np.mean(distribuicoes["Binomial"]["dados"])
mediana_bin = np.median(distribuicoes["Binomial"]["dados"])
desvio_padrao_bin = np.std(distribuicoes["Binomial"]["dados"])
variancia_bin = np.var(distribuicoes["Binomial"]["dados"])
distribuicoes["Binomial"]["estatisticas"] = {"média": media_bin, "mediana": mediana_bin, "desvio_padrão":desvio_padrao_bin, "variância": variancia_bin}
```

```python
pd.DataFrame({"Normal": distribuicoes["Normal"]["estatisticas"],
              "Exponencial" : distribuicoes["Exponencial"]["estatisticas"],
              "Uniforme": distribuicoes["Uniforme"]["estatisticas"],
              "Poisson": distribuicoes["Poisson"]["estatisticas"],
              "Binomial": distribuicoes["Binomial"]["estatisticas"]
              })
```

```python
min_normal = np.min(distribuicoes["Normal"]["dados"])
max_normal = np.max(distribuicoes["Normal"]["dados"])
quartis_normal = np.percentile(distribuicoes["Normal"]["dados"],q=(25,50,75))
distribuicoes["Normal"]["estatisticas"]["máximo"] = max_normal
```
