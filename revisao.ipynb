**Objetivo:**

## Passo 1 – Criar o dicionário principal
As chaves devem ser nomes de distribuições: `"Normal"`, `"Exponencial"`, `"Uniforme"`  
Os valores devem ser dicionários contendo:  
- `"params"` → parâmetros da distribuição  
- `"dados"` → inicialmente `None`  
- `"estatisticas"` → inicialmente `None`  

Exemplo de estrutura:

    distribuicoes = {
        "Normal": {"params": {"loc": 0, "scale": 1}, "dados": None, "estatisticas": None},
        "Exponencial": {"params": {"scale": 2}, "dados": None, "estatisticas": None},
        "Uniforme": {"params": {"low": -1, "high": 1}, "dados": None, "estatisticas": None}
    }

---

```python
distribuicoes = {
    "Normal": {"params": {"loc": 0, "scale": 1}, "dados": None, "estatisticas": None},
    "Exponencial": {"params": {"scale": 2}, "dados": None, "estatisticas": None},
    "Uniforme": {"params": {"low": -1, "high": 1}, "dados": None, "estatisticas": None}
}
```

## Passo 2 – Importar bibliotecas
Importe as bibliotecas:  
- `numpy` para simulação  
- `pandas` para organização dos resultados  
- `matplotlib` para visualização  

---

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Passo 3 – Gerar os dados
Para cada distribuição, gere uma amostra de tamanho **1000** usando `numpy.random` e salve em `"dados"`.  

---

```python
distribuicoes["Normal"]["dados"] = np.random.normal(loc=0, scale=1, size=1000)
distribuicoes["Exponencial"]["dados"] = np.random.exponential(scale=2, size=1000)
distribuicoes["Uniforme"]["dados"] = np.random.uniform(low=-1, high=1, size=1000)
```

Aqui eu apenas gerei os dados das respectivas distribuicoes e os adicionei em "dados"

## Passo 4 – Calcular estatísticas básicas
Para cada distribuição, calcule:  
- Média  
- Mediana  
- Variância  
- Desvio padrão  
Armazene em um dicionário dentro da chave `"estatisticas"`.  

### Calculando para a Normal

```python
media_normal = np.mean(distribuicoes["Normal"]["dados"])
mediana_normal = np.median(distribuicoes["Normal"]["dados"])
desvio_padrao_normal = np.std(distribuicoes["Normal"]["dados"])
variancia_normal = np.var(distribuicoes["Normal"]["dados"])
distribuicoes["Normal"]["estatisticas"] = {"média": media_normal, "mediana": mediana_normal, "desvio_padrão":desvio_padrao_normal, "variância": variancia_normal}

```

### Calculando para a Exponencial

```python
media_exp = np.mean(distribuicoes["Exponencial"]["dados"])
mediana_exp = np.median(distribuicoes["Exponencial"]["dados"])
desvio_padrao_exp = np.std(distribuicoes["Exponencial"]["dados"])
variancia_exp = np.var(distribuicoes["Exponencial"]["dados"])
distribuicoes["Exponencial"]["estatisticas"] = {"média": media_exp, "mediana": mediana_exp, "desvio_padrão":desvio_padrao_exp, "variância": variancia_exp}
```

### Calculando para a Uniforme


```python
media_uni = np.mean(distribuicoes["Uniforme"]["dados"])
mediana_uni = np.median(distribuicoes["Uniforme"]["dados"])
desvio_padrao_uni = np.std(distribuicoes["Uniforme"]["dados"])
variancia_uni = np.var(distribuicoes["Uniforme"]["dados"])
distribuicoes["Uniforme"]["estatisticas"] = {"média": media_uni, "mediana": mediana_uni, "desvio_padrão":desvio_padrao_uni, "variância": variancia_uni}
```

Aqui eu calculei separadamente as estatisticas descritivas coloquei elas em "estatisticas" em suas respectivas distribuicoes

---

## Passo 5 – Visualizar os dados
Para cada distribuição, faça um **histograma** mostrando a forma da distribuição.  

```python
plt.hist(distribuicoes["Normal"]["dados"])
```

```python
plt.hist(distribuicoes["Exponencial"]["dados"])
```

```python
plt.hist(distribuicoes["Uniforme"]["dados"])
```

Aqui eu plotei histogramas com os dados de cada distribuição, que no caso da normal por exemplo estão em distribuicoes["Normal"]["dados"]

---

## Passo 6 – Organizar resultados em Pandas
Crie um `DataFrame` contendo as estatísticas de todas as distribuições.  

```python
pd.DataFrame({"Normal": distribuicoes["Normal"]["estatisticas"],
              "Exponencial" : distribuicoes["Exponencial"]["estatisticas"],
              "Uniforme": distribuicoes["Uniforme"]["estatisticas"]})
```

Aqui eu criei um DataFrame (uma tabela) que exibe as estatísticas descritivas que eu calcule para cada distribuição, para assim poder comparar os valores amostrais com os valores globais (populacionais)

---

## Passo 7 – Adicionar novas distribuições
- Inclua mais uma distribuição, como `"Poisson"` ou `"Binomial"`.  
- Gere dados usando `numpy.random`.  
- Calcule as estatísticas básicas e atualize o `DataFrame`.  

```python
distribuicoes["Poisson"] = {"params": {"lam": 5}, "dados" : None, "estatisticas" : None }
distribuicoes["Binomial"] = {"params": {"n": 1, "p" : 0.5}, "dados" : None, "estatisticas" : None}

distribuicoes["Poisson"]["dados"] = np.random.poisson(lam=5,size=1000)
distribuicoes["Binomial"]["dados"] = np.random.binomial(n=1,p=0.5,size=1000)
```

Aqui eu adicionei as novas distribuicoes no nosso dicionario "distribuicoes", criei seus valores tambem (que é um outro dicionario). Também gerei os dados e adicionei eles em "dados" de cada distribuicao

```python
media_pois = np.mean(distribuicoes["Poisson"]["dados"])
mediana_pois = np.median(distribuicoes["Poisson"]["dados"])
desvio_padrao_pois = np.std(distribuicoes["Poisson"]["dados"])
variancia_pois = np.var(distribuicoes["Poisson"]["dados"])
distribuicoes["Poisson"]["estatisticas"] = {"média": media_pois, "mediana": mediana_pois, "desvio_padrão":desvio_padrao_pois, "variância": variancia_pois}

media_bin = np.mean(distribuicoes["Binomial"]["dados"])
mediana_bin = np.median(distribuicoes["Binomial"]["dados"])
desvio_padrao_bin = np.std(distribuicoes["Binomial"]["dados"])
variancia_bin = np.var(distribuicoes["Binomial"]["dados"])
distribuicoes["Binomial"]["estatisticas"] = {"média": media_bin, "mediana": mediana_bin, "desvio_padrão":desvio_padrao_bin, "variância": variancia_bin}
```

Aqui eu calculei as estatísticas descritivas da poisson e da binomial e as adicionei em "estatisticas"

```python
pd.DataFrame({"Normal": distribuicoes["Normal"]["estatisticas"],
              "Exponencial" : distribuicoes["Exponencial"]["estatisticas"],
              "Uniforme": distribuicoes["Uniforme"]["estatisticas"],
              "Poisson": distribuicoes["Poisson"]["estatisticas"],
              "Binomial": distribuicoes["Binomial"]["estatisticas"]
              })
```

Aqui eu atualizei a minha DataFrame com as informacoes sobre as novas distribuicoes

---

## Passo 8 – Estatísticas adicionais
Para cada distribuição, calcule também:  
- Mínimo (`np.min`)  
- Máximo (`np.max`)  
- Quartis (`np.percentile`)  

Atualize o campo `"estatisticas"` e o `DataFrame` com essas novas colunas.  

### Normal

```python
min_normal = np.min(distribuicoes["Normal"]["dados"])
max_normal = np.max(distribuicoes["Normal"]["dados"])
quartis_normal = np.percentile(distribuicoes["Normal"]["dados"],q=(25,50,75))
distribuicoes["Normal"]["estatisticas"]["máximo"] = max_normal
distribuicoes["Normal"]["estatisticas"]["mínimo"] = min_normal
distribuicoes["Normal"]["estatisticas"]["quartis"] = quartis_normal
```

```python

nomes_distribuicoes = ["Exponencial", "Uniforme", "Poisson", "Binomial"]

for nome in nomes_distribuicoes:
    dados = distribuicoes[nome]["dados"]

    min_val = np.min(dados)
    max_val = np.max(dados)
    quartis = np.percentile(dados, q=(25, 50, 75))

    distribuicoes[nome]["estatisticas"]["mínimo"] = min_val
    distribuicoes[nome]["estatisticas"]["máximo"] = max_val
    distribuicoes[nome]["estatisticas"]["quartis"] = quartis
```

Aqui para tentar deixar um pouco mais pratico, criei um loop para adicionar as novas descritivas no dicionario "estatisticas", que vai pegando distribuicao por distribuicao

```python
pd.DataFrame({"Normal": distribuicoes["Normal"]["estatisticas"],
              "Exponencial" : distribuicoes["Exponencial"]["estatisticas"],
              "Uniforme": distribuicoes["Uniforme"]["estatisticas"],
              "Poisson": distribuicoes["Poisson"]["estatisticas"],
              "Binomial": distribuicoes["Binomial"]["estatisticas"]
              })
```

Aqui, eu apenas fiz o DataFrame novamente com as informacoes atualizadas

---

## Passo 9 – Comparar distribuições graficamente
Faça um **boxplot** com todas as distribuições no mesmo gráfico, com título e rótulos adequados.  

---

```python
plt.boxplot(distribuicoes["Normal"]["dados"])
plt.boxplot(distribuicoes["Exponencial"]["dados"])
plt.boxplot(distribuicoes["Uniforme"]["dados"])
plt.boxplot(distribuicoes["Poisson"]["dados"])
plt.boxplot(distribuicoes["Binomial"]["dados"])
```

```python
# 2. Agrupar os dados em uma lista
dados = [distribuicoes["Normal"]["dados"],
        distribuicoes["Exponencial"]["dados"],
        distribuicoes["Uniforme"]["dados"],
        distribuicoes["Poisson"]["dados"],
        distribuicoes["Binomial"]["dados"]]

# 3. Plotar os boxplots juntos
fig, ax = plt.subplots()
ax.boxplot(dados, tick_labels=['Normal', 'Exponencial', 'Uniforme', 'Poisson','Binomial'])

# 4. Personalização do gráfico
ax.set_title('Comparação das distribuições')
ax.set_ylabel('Valores')
ax.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

Aqui eu pesquisei no gemini como gerar varios boxplots um do lado do outro em um mesmo gráfico
